# Exploratory Analysis of 3D Glioma Radiomic Features
Initial inspection of class balance, feature distributions across molecular phenotypes, and inter-feature correlation (collinearity) from `data/processed/radiomic_features.csv`.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid", context="notebook")
features_path = Path("../data/processed/radiomic_features.csv")
df = pd.read_csv(features_path)
print(f"Shape: {df.shape[0]} subjects x {df.shape[1]} columns")
print("Missing values:", int(df.isna().sum().sum()))
print("\nClass distribution (label):")
print(df["label"].value_counts(dropna=False))
df.head()
```

```python
num_cols = [c for c in df.select_dtypes(include=[np.number]).columns if c != "label"]
top_feats = num_cols[:6]
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.ravel(), top_feats):
    sns.boxplot(data=df, x="label", y=col, ax=ax)
    ax.set_title(col, fontsize=10)
plt.tight_layout()
plt.show()
```

```python
corr = df[num_cols].corr(method="spearman")
plt.figure(figsize=(11, 9))
sns.heatmap(corr, cmap="coolwarm", center=0, square=True, cbar_kws={"shrink": 0.8})
plt.title("Spearman Correlation Matrix Across 3D Radiomic Features")
plt.tight_layout()
plt.show()
```
